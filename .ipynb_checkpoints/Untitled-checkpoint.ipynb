```python
import re
import requests
from bs4 import BeautifulSoup

# 1. Usar la URL con los scripts del jugador
url = "https://www.tennisabstract.com/cgi-bin/player-classic.cgi?p=AlexDeMinaur"

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

# 2. Extraer usando .text en lugar de .string
script_text = ""
for script in soup.find_all("script"):
    content = script.text or ""
    if "ychoices" in content:
        script_text = content
        break

columnas = []
variables_encabezado = ["ychoices", "tchoices", "cchoices", "rchoices"]

if script_text:
    for var in variables_encabezado:
        # Expresión regular ajustada para capturar el primer elemento dentro del array de la variable
        match = re.search(r'' + var + r'\s*=\s*\[\s*["\']([^"\']+)["\']', script_text)
        if match:
            columnas.append(match.group(1))

print("Títulos de la tabla encontrados:")
print(columnas)
```

```python
import re
import json
import requests
from bs4 import BeautifulSoup

# URL exacta del perfil del tenista (donde vive la base de datos de partidos)
url = "https://www.tennisabstract.com/cgi-bin/player.cgi?p=AlexDeMinaur"

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

# 1. Localizar el script que contiene la base de datos del jugador
script_datos = ""
for script in soup.find_all("script"):
    content = script.text or ""
    if "var matches" in content or "theScores" in content:
        script_datos = content
        break

# 2. Si las variables se cargan desde el documento complementario -classic (Truco de extracción extendido)
if not script_datos:
    # A veces el backend de la web delega la matriz completa a su versión clásica síncrona
    url_classic = "https://www.tennisabstract.com/cgi-bin/player-classic.cgi?p=AlexDeMinaur"
    response = requests.get(url_classic, headers=headers)
    soup = BeautifulSoup(response.text, "html.parser")
    for script in soup.find_all("script"):
        content = script.text or ""
        if "matches" in content or "ochoices" in content:
            script_datos = content
            break

# 3. Parsear las filas de los encuentros 
print("--- Procesando registros de la tabla ---")
if script_datos:
    # Buscamos estructuras de filas estructuradas dentro del JavaScript
    # Tennis Abstract organiza los strings de partidos separándolos por comas o dentro de un diccionario estructurado
    partidos_crudos = re.findall(r'\[["\'](.*?)["\']\]', script_datos)
    
    if partidos_crudos:
        for idx, partido in enumerate(partidos_crudos[:15]): # Limitado a los primeros 15 resultados para la muestra
            print(f"Partido {idx+1}: {partido}")
    else:
        # Fallback alternativo: Extraer directamente todos los textos limpios de los rivales listados
        match_rivales = re.search(r'ochoices\s*=\s*\[(.*?)\];', script_datos, re.DOTALL)
        if match_rivales:
            rivales = re.findall(r'["\']([^"\']+)["\']', match_rivales.group(1))
            print("Rivales encontrados en la base de datos de la tabla:")
            print(rivales[:10]) # Imprime los primeros 10 contrincantes de las filas
else:
    print("No se pudo acceder a la matriz interna. Verifica que el parámetro 'p=AlexDeMinaur' esté escrito tal cual.")
```
